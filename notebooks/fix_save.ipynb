```python
import os

path = r'D:\E-Commerce-Customer-Analytics\data\processed'
print("Folder exists:", os.path.exists(path))

# Create folder if missing
os.makedirs(path, exist_ok=True)
print("Folder created/confirmed!")

# List what's inside
files = os.listdir(path)
print("Files in processed folder:", files)
```

```python
import pandas as pd

print("Loading raw data...")
df = pd.read_csv(
    r'D:\E-Commerce-Customer-Analytics\data\Online Retail.csv',
    encoding='latin-1',
    dtype={'CustomerID': str}
)
print(f"Loaded: {len(df):,} rows")

# Basic cleaning
df = df.dropna(subset=['CustomerID'])
df = df[df['Quantity'] > 0]
df = df[df['UnitPrice'] > 0]
df = df[~df['InvoiceNo'].astype(str).str.startswith('C')]

# New columns
df['TotalPrice'] = df['Quantity'] * df['UnitPrice']
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'])
df['YearMonth'] = df['InvoiceDate'].dt.to_period('M').astype(str)
df['Hour'] = df['InvoiceDate'].dt.hour
df['DayOfWeek'] = df['InvoiceDate'].dt.day_name()
df['Month'] = df['InvoiceDate'].dt.month

print(f"Cleaned rows: {len(df):,}")
print(f"Columns: {list(df.columns)}")
```

```python
save_path = r'D:\E-Commerce-Customer-Analytics\data\processed\cleaned_retail.csv'

df.to_csv(save_path, index=False)

# Verify it saved
if os.path.exists(save_path):
    size = os.path.getsize(save_path)
    print(f"✅ SUCCESS! File saved!")
    print(f"File size: {size/1024/1024:.1f} MB")
    print(f"Location: {save_path}")
else:
    print("❌ FAILED - file not saved")
```
